```python
import pandas as pd
```

```python
url = "https://www.nps.gov/articles/000/historic-listing-of-nps-park-codes.htm"

tables = pd.read_html(url)
len(tables)
```

```python
df = tables[0]
df.head()
```

```python
df.columns = df.iloc[0]
df = df.iloc[1:].reset_index(drop=True)
df.head()
```

```python
df.to_csv('park_names.csv', index=False)
```
